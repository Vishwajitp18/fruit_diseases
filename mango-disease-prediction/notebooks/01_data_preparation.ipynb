Check image dimensions

```python
from PIL import Image

dimension_counts = {}

for class_name in classes:

    class_dir = MANGO_DIR / class_name

    images = [
        path
        for path in class_dir.rglob("*")
        if path.is_file()
        and path.suffix.lower() in image_extensions
    ]

    for image_path in images:

        try:
            with Image.open(image_path) as img:
                size = img.size

            dimension_counts[size] = dimension_counts.get(size, 0) + 1

        except Exception as e:
            print("Could not read:", image_path)
            print("Error:", e)

print("Number of unique image dimensions:", len(dimension_counts))

print("\nMost common dimensions:")

for size, count in sorted(
    dimension_counts.items(),
    key=lambda x: x[1],
    reverse=True
)[:20]:
    print(f"{size}: {count}")
```

Check corrupted images

```python
from PIL import Image

corrupted_images = []

for class_name in classes:

    class_dir = MANGO_DIR / class_name

    images = [
        path
        for path in class_dir.rglob("*")
        if path.is_file()
        and path.suffix.lower() in image_extensions
    ]

    for image_path in images:

        try:
            with Image.open(image_path) as img:
                img.verify()

        except Exception as e:
            corrupted_images.append((image_path, str(e)))


print("Corrupted images:", len(corrupted_images))

if corrupted_images:

    print("\nExamples:")

    for image_path, error in corrupted_images[:10]:
        print(image_path)
        print(error)
```

```python
IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".webp"
}

dataset_counts = {}

for class_name in CLASSES:

    class_directory = MANGO_DIR / class_name

    images = [
        path
        for path in class_directory.rglob("*")
        if path.is_file()
        and path.suffix.lower() in IMAGE_EXTENSIONS
    ]

    dataset_counts[class_name] = len(images)


print("DATASET DISTRIBUTION")
print("=" * 50)

total_images = 0

for class_name, count in dataset_counts.items():

    print(f"{class_name:20s}: {count}")

    total_images += count

print("=" * 50)
print("Total images:", total_images)
```

```python
records = []

for label, class_name in enumerate(CLASSES):

    class_directory = MANGO_DIR / class_name

    images = [
        path
        for path in class_directory.rglob("*")
        if path.is_file()
        and path.suffix.lower() in IMAGE_EXTENSIONS
    ]

    for image_path in images:

        records.append({
            "image_path": str(image_path),
            "class_name": class_name,
            "label": label
        })


df = pd.DataFrame(records)

print("Number of records:", len(df))

df.head()
```

```python
PROJECT_ROOT = Path(
    "/content/drive/MyDrive/project-1/mango-disease-prediction"
)

MANGO_DIR = PROJECT_ROOT / "data" / "raw" / "mango"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

CLASSES = [
    "Anthracnose",
    "Bacterial Canker",
    "Cutting Weevil",
    "Die Back",
    "Gall Midge",
    "Healthy",
    "Powdery Mildew",
    "Sooty Mould"
]

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".webp"
}

print("Project:", PROJECT_ROOT)
print("Dataset:", MANGO_DIR)
```

Create metadata

```python
records = []

for label, class_name in enumerate(CLASSES):

    class_dir = MANGO_DIR / class_name

    images = [
        path
        for path in class_dir.rglob("*")
        if path.is_file()
        and path.suffix.lower() in IMAGE_EXTENSIONS
    ]

    print(f"{class_name:20s}: {len(images)}")

    for image_path in images:

        records.append({
            "image_path": str(image_path),
            "class_name": class_name,
            "label": label
        })


df = pd.DataFrame(records)

print("\nTotal images:", len(df))
```

```python
class_counts = df["class_name"].value_counts()

print(class_counts)
```

```python
sample_df = df.sample(
    n=min(16, len(df)),
    random_state=42
)

fig, axes = plt.subplots(
    4, 4,
    figsize=(14, 14)
)

for ax, (_, row) in zip(
    axes.flatten(),
    sample_df.iterrows()
):

    image = Image.open(row["image_path"])

    ax.imshow(image)
    ax.set_title(row["class_name"])
    ax.axis("off")


plt.tight_layout()
plt.show()
```

```python
dimension_records = []

sample_size = min(200, len(df))

dimension_sample = df.sample(
    n=sample_size,
    random_state=42
)

for _, row in dimension_sample.iterrows():

    try:

        with Image.open(row["image_path"]) as image:

            width, height = image.size

        dimension_records.append({
            "width": width,
            "height": height
        })

    except Exception as e:

        print("Could not read:", row["image_path"])
        print(e)


dimension_df = pd.DataFrame(dimension_records)

print("Images inspected:", len(dimension_df))

dimension_df.value_counts().head(20)
```

Check corrupted images

```python
corrupted_images = []

sample_size = min(200, len(df))

validation_sample = df.sample(
    n=sample_size,
    random_state=42
)

for _, row in validation_sample.iterrows():

    image_path = row["image_path"]

    try:

        with Image.open(image_path) as image:
            image.verify()

    except Exception as e:

        corrupted_images.append({
            "image_path": image_path,
            "error": str(e)
        })


print("Images checked:", sample_size)
print("Corrupted images:", len(corrupted_images))
```
